```python
import subprocess

# Install Weka and dependencies
!sudo apt-get install weka

# Define the CSV file paths for training and testing
training_csv_path = '/content/sample_data/CNS_Train_YJ_PC100genes.arff'
testing_csv_path = '/content/sample_data/CNS_Test_YJ_PC100genes.arff'

# Convert training CSV to ARFF format using Weka's CSVLoader
!java -cp /usr/share/java/weka.jar weka.core.converters.CSVLoader $training_csv_path > training_data.arff

# Convert testing CSV to ARFF format using Weka's CSVLoader
!java -cp /usr/share/java/weka.jar weka.core.converters.CSVLoader $testing_csv_path > testing_data.arff

# Define the Weka command for training the J48 classifier
train_command = f"java -cp /usr/share/java/weka.jar weka.classifiers.trees.J48 -t training_data.arff -d j48_model.model -no-cv"

# Run the Weka command for training
train_result = subprocess.run(train_command, shell=True, stdout=subprocess.PIPE, stderr=subprocess.PIPE, text=True)

# Print the training result
print("Training Result:\n", train_result.stdout)
print(train_result.stderr)

# Define the Weka command for testing the J48 classifier
test_command = f"java -cp /usr/share/java/weka.jar weka.classifiers.trees.J48 -l j48_model.model -T testing_data.arff"

# Run the Weka command for testing
test_result = subprocess.run(test_command, shell=True, stdout=subprocess.PIPE, stderr=subprocess.PIPE, text=True)

# Print the testing result
print("\nTesting Result:\n", test_result.stdout)
print(test_result.stderr)

```

```python
import urllib.request
from google.colab import files
import subprocess

# Install Weka and dependencies
subprocess.run(['sudo', 'apt-get', 'install', 'weka'])

# Define the URL for libsvm.jar from the official repository
libsvm_jar_url = 'https://www.csie.ntu.edu.tw/~cjlin/libsvm/libsvm.jar'
urllib.request.urlretrieve(libsvm_jar_url, 'libsvm.jar')

# Upload training ARFF file
print("Upload the training ARFF file:")
uploaded_training = files.upload()
training_arff_path = list(uploaded_training.keys())[0]

# Upload testing ARFF file
print("\nUpload the testing ARFF file:")
uploaded_testing = files.upload()
testing_arff_path = list(uploaded_testing.keys())[0]

# Define the Weka command for training the LibSVM classifier
train_command = [
    'java', '-cp', '/usr/share/java/weka.jar:libsvm.jar',
    'weka.classifiers.functions.LibSVM', '-t', training_arff_path, '-d', 'svm_model.model'
]

# Run the Weka command for training
train_result = subprocess.run(train_command, capture_output=True, text=True)

# Print the training result
print("Training Result:\n", train_result.stdout)
print(train_result.stderr)

# Define the Weka command for testing the LibSVM classifier
test_command = [
    'java', '-cp', '/usr/share/java/weka.jar:libsvm.jar',
    'weka.classifiers.functions.LibSVM', '-l', 'svm_model.model', '-T', testing_arff_path
]

# Run the Weka command for testing
test_result = subprocess.run(test_command, capture_output=True, text=True)

# Print the testing result
print("\nTesting Result:\n", test_result.stdout)
print(test_result.stderr)
```
